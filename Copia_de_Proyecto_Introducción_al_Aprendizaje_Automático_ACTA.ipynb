```python
# Detección de Outliers usando el método IQR

for col in columnas_num:
    Q1 = df_ev[col].quantile(0.25)
    Q3 = df_ev[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_ev[(df_ev[col] < lower_bound) | (df_ev[col] > upper_bound)]

    print(f"\n--- Outliers para la columna: {col.replace('_', ' ').title()} ---")
    print(f"Total de outliers: {len(outliers)}")
    if not outliers.empty:
        print("Primeros 5 outliers:")
        display(outliers[[col]].head())
```

El análisis de outliers y estadístico se realiza para las columnas numéricas. Se ha utilizado el método del rango intercuartílico (IQR) para identificar valores que se encuentran significativamente fuera del rango de la mayoría de los datos. Esto nos permite tener una visión de la dispersión y la presencia de valores extremos que podrían requerir un tratamiento especial.

Dado que los datos que presentan outliers son importantes para el análisis siguiente se mantienen los valores.

```python
# Proporción de vehículos por tipo de EV

# Definir variables para la agrupación

evt = lambda x: 'Plug-in Hybrid Electric Vehicle (PHEV)' if x == 'Plug-in Hybrid Electric Vehicle (PHEV)' else 'Battery Electric Vehicle (BEV)'
df_ev['agrupacion_evt'] = df_ev['Electric Vehicle Type'].apply(evt)
pie_data = df_ev['agrupacion_evt'].value_counts()


# Realizar gráfico de dona

colors_pie = ['#4285F4', '#D3D3D3'] # Azul = "PHEV", Gris = "BEV"
explode_val = 0.05 # Valor de separación


# Gráfico de pastel

plt.figure(figsize=(6, 6))
wedges, texts, autotexts = plt.pie(pie_data,
                                   labels=pie_data.index,
                                   autopct='%1.1f%%',
                                   startangle=90,
                                   colors=colors_pie,
                                   textprops={'fontsize': 7, 'fontweight': 'bold'},
                                   pctdistance=0.85, # Mueve el porcentaje hacia afuera
                                   explode=(explode_val, 0)) # Resalta ligeramente la porción principal


# Círculo blanco central (estilo moderno)

theta1, theta2 = wedges[0].theta1, wedges[0].theta2
center_angle = np.deg2rad((theta1 + theta2) / 2) # Obtención el ángulo medio

cx = explode_val * np.cos(center_angle) # Se Calcula nueva posición con trigonometría
cy = explode_val * np.sin(center_angle)

centre_circle = plt.Circle((cx, cy), 0.70, fc='white') # Se dibuja el circulo
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Proporción de Tipo de EV de Videos', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

Los vehículos totalmente eléctricos (BEV) representan aproximadamente el 78.3% del mercado, mientras que los híbridos enchufables (PHEV) constituyen el 21.7%. Esto indica una clara preferencia del mercado por tecnologías completamente eléctricas.

**5. Aprendizaje Automático**

Se realiza la predicción del tamaño del mercado en los años 2025-2030

```python
# Regresión lineal

import tensorflow as tf

xs = market_year['Model Year']
ys = market_year['vehicles']

regresion_lineal = np.polyfit(xs, ys, 1)
# y = b0x + b1
# beta --> b0 y alfa --> b1
beta, alfa = regresion_lineal
print(f'y = {round(beta, 1)}x + {round(alfa, 1)}')
```

```python
plt.figure(figsize=(10, 6)) # Adjust figure size for better readability
plt.plot(xs, ys, 'o', color='green', label='Datos')
plt.plot(xs, beta*xs + alfa, color='blue', label='Regresión Lineal')
plt.title('Predicción del mercado EV (Regresión Lineal)', fontsize=14)
plt.xlabel('Año del Modelo (xs)')
plt.ylabel('Vehículos Registrados (ys)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7) # Add a grid for better data interpretation
plt.show()
```

```python
# Predicción futura utilizando la regresión lineal (polyfit)

# Definir los años futuros para la predicción
future_years_lr = pd.DataFrame({
    'Model Year': np.arange(2025, 2031) # Generar años del 2025 al 2030
})

# Calcular las predicciones usando la ecuación lineal: y = beta * x + alfa
# 'beta' y 'alfa' fueron obtenidos de regresion_lineal = np.polyfit(xs, ys, 1)
future_pred_lr = beta * future_years_lr['Model Year'] + alfa

# Crear un DataFrame para mostrar las predicciones
future_df_lr = pd.DataFrame({
    "Year": future_years_lr["Model Year"],
    "Predicted_Vehicles": future_pred_lr.astype(int)
})

print("Predicción de Vehículos Eléctricos con Regresión Lineal para años futuros:")
display(future_df_lr)

# Visualizar los datos originales y la predicción de la regresión lineal
plt.figure(figsize=(10,6))

plt.scatter(xs, ys, label="Datos Reales", color='green')
plt.plot(
    future_years_lr['Model Year'],
    future_pred_lr,
    color="red",
    linewidth=3,
    label="Predicción de Regresión Lineal"
)

# Añadir la línea de regresión para los datos históricos
plt.plot(xs, beta * xs + alfa, color='blue', linestyle='--', label='Línea de Regresión (Histórico)')

plt.title("Predicción del Mercado EV con Regresión Lineal", fontsize=14)
plt.xlabel("Año del Modelo")
plt.ylabel("Vehículos Registrados")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

Se aprecia que el modelo no estima adecuadamente en los últimos años, por el cambio de tendencia en el año 2024, por lo que se opta por eliminar ese año y rehacer el modelo de regresión.

```python
# eliminar años 2024

df_filtrado = df_ev[df_ev['Model Year'] != 2024]
```

```python
# Regresión lineal sin año 2024

import tensorflow as tf

market_year = (
    df_filtrado.groupby("Model Year")
      .size()
      .reset_index(name="vehicles")
)

xs = market_year['Model Year']
ys = market_year['vehicles']

regresion_lineal = np.polyfit(xs, ys, 1)
# y = b0x + b1
# beta --> b0 y alfa --> b1
beta, alfa = regresion_lineal
print(f'y = {round(beta, 1)}x + {round(alfa, 1)}')
```

```python
# This cell will now plot the Neural Network's predictions instead of Linear Regression

# Define future years for prediction (same as before)
future_years_nn_plot = pd.DataFrame({
    'Model Year': np.arange(2025, 2031) # Generar años del 2025 al 2030
})

# Scale future years for prediction using the same scaler
future_years_scaled = xs_scaler.transform(future_years_nn_plot['Model Year'].to_numpy().reshape(-1, 1))

# Make predictions with the trained neural network model
predicted_scaled_nn = modelo.predict(future_years_scaled)

# Inverse transform the scaled predictions to original scale
predicted_vehicles_nn = ys_scaler.inverse_transform(predicted_scaled_nn).flatten()

plt.figure(figsize=(10, 6))
plt.plot(xs, ys, 'o', color='green', label='Datos Reales (sin 2024)')
plt.plot(
    future_years_nn_plot['Model Year'],
    predicted_vehicles_nn,
    color='red',
    linewidth=3,
    label='Predicción de Red Neuronal'
)

# Plot the neural network's fit on the historical data (optional, but good for context)
historical_pred_scaled = modelo.predict(xs_scaled)
historical_pred_nn = ys_scaler.inverse_transform(historical_pred_scaled).flatten()
plt.plot(xs, historical_pred_nn, color='blue', linestyle='--', label='Ajuste de Red Neuronal (Histórico)')

plt.title('Predicción del Mercado EV con Red Neuronal', fontsize=14)
plt.xlabel('Año del Modelo')
plt.ylabel('Vehículos Registrados')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Display the numerical predictions as a table for clarity
future_df_nn_plot = pd.DataFrame({
    "Year": future_years_nn_plot["Model Year"],
    "Predicted_Vehicles": predicted_vehicles_nn.astype(int)
})
print("Predicción de Vehículos Eléctricos con Red Neuronal para años futuros:")
display(future_df_nn_plot)
```

```python
# Predicción futura utilizando la regresión lineal (polyfit)

# Definir los años futuros para la predicción
future_years_lr = pd.DataFrame({
    'Model Year': np.arange(2025, 2031) # Generar años del 2025 al 2030
})

# Calcular las predicciones usando la ecuación lineal: y = beta * x + alfa
# 'beta' y 'alfa' fueron obtenidos de regresion_lineal = np.polyfit(xs, ys, 1)
future_pred_lr = beta * future_years_lr['Model Year'] + alfa

# Crear un DataFrame para mostrar las predicciones
future_df_lr = pd.DataFrame({
    "Year": future_years_lr["Model Year"],
    "Predicted_Vehicles": future_pred_lr.astype(int)
})

print("Predicción de Vehículos Eléctricos con Regresión Lineal para años futuros:")
display(future_df_lr)

# Visualizar los datos originales y la predicción de la regresión lineal
plt.figure(figsize=(10,6))

plt.scatter(xs, ys, label="Datos Reales", color='green')
plt.plot(
    future_years_lr['Model Year'],
    future_pred_lr,
    color="red",
    linewidth=3,
    label="Predicción de Regresión Lineal"
)

# Añadir la línea de regresión para los datos históricos
plt.plot(xs, beta * xs + alfa, color='blue', linestyle='--', label='Línea de Regresión (Histórico)')

plt.title("Predicción del Mercado EV con Regresión Lineal", fontsize=14)
plt.xlabel("Año del Modelo")
plt.ylabel("Vehículos Registrados")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

Se aprecia una mejora en la estimación, sin embargo, por las características de la curva de crecimiento, una regresión lineal no es adecuada para realizar la estimación.

```python
# Red neuronal - Se solicitó ayuda de la IA para determinar el tipo de red adecuado para los datos elegidos

import numpy as np
from sklearn.preprocessing import StandardScaler

# xs and ys are already updated globally from cHI5ASlM7A6H (without 2024 data)
xs_array = xs.to_numpy().reshape(-1, 1)
ys_array = ys.to_numpy().reshape(-1, 1)

# Crear escaladores para xs y ys
xs_scaler = StandardScaler()
ys_scaler = StandardScaler()

# Escalar xs
xs_scaled = xs_scaler.fit_transform(xs_array)

# Escalar ys
ys_scaled = ys_scaler.fit_transform(ys_array)

# Define a more complex model
model_nn = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),        # Input layer
    tf.keras.layers.Dense(64, activation='relu'), # Hidden layer 1
    tf.keras.layers.Dense(32, activation='relu'), # Hidden layer 2
    tf.keras.layers.Dense(1)           # Output layer (linear activation for regression)
])

# Compile with Adam optimizer
model_nn.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.005), loss='mse')

model_nn.summary()

# Assign to 'modelo' for consistency with other cells that might use it
modelo = model_nn
```

```python
print('Comenzando entrenamiento...')
# Usar xs_scaled y ys_scaled para el entrenamiento
historial = modelo.fit(xs_scaled, ys_scaled, epochs=1000, verbose=False) # Epocas incrementadas
print('Entrenamiento finalizado')

# Mostrar la pérdida final
print(f"Pérdida final del entrenamiento: {historial.history['loss'][-1]:.4f}")

# Inspeccionar la última capa.
final_layer_weights = modelo.layers[-1].get_weights()
print(f"Pesos de la capa final: {final_layer_weights[0][0][0]:.4f}")
print(f"Sesgo de la capa final: {final_layer_weights[1][0]:.4f}")
```

Se incrementa el número de épocas para minimizar la pérdida, en este proyecto se consideran 1000 épocas.

```python
# Graficar el historial de pérdidas
plt.figure(figsize=(10, 6))
plt.plot(historial.history['loss'], label='Pérdida de Entrenamiento')
plt.title('Comportamiento de la Pérdida durante el Entrenamiento')
plt.xlabel('Época')
plt.ylabel('Pérdida (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Evaluación de la red neuronal
# ----------------------------------------------------------

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt

# Emplear valores escalados
X_train_scaled, X_test_scaled, y_train_scaled, y_test_scaled = train_test_split(
    xs_scaled,
    ys_scaled,
    test_size=0.3,
    random_state=42
)

# Predecir empleando el set de prueba
pred_scaled = modelo.predict(X_test_scaled)

# Realizar la transformación inversa para estimar métricas de error
y_test_original = ys_scaler.inverse_transform(y_test_scaled)
pred_original = ys_scaler.inverse_transform(pred_scaled)

print("\nMétricas de la Red Neuronal (en escala original):")
print(f"MAE: {mean_absolute_error(y_test_original, pred_original):.2f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_original, pred_original)):.2f}")
print(f"R²: {r2_score(y_test_original, pred_original):.2f}")

# Visualización de las predicciones vs. valores reales en el conjunto de prueba
plt.figure(figsize=(10, 6))
plt.scatter(y_test_original, pred_original, alpha=0.6, color='purple')
plt.plot([y_test_original.min(), y_test_original.max()], [y_test_original.min(), y_test_original.max()], 'r--', lw=2, label='Predicción Perfecta')
plt.title('Predicciones de la Red Neuronal vs. Valores Reales (Test Set)', fontsize=14)
plt.xlabel('Valores Reales (Vehículos Registrados)')
plt.ylabel('Predicciones (Vehículos Registrados)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Visualización del modelo ajustado en el contexto de los datos completos
plt.figure(figsize=(12, 7))
plt.scatter(xs, ys, label="Datos Reales (sin 2024)", color='green', alpha=0.7)

# Predicciones del modelo en los datos originales (inverso de xs_scaled)
hist_pred_scaled = modelo.predict(xs_scaled)
hist_pred_original = ys_scaler.inverse_transform(hist_pred_scaled)
plt.plot(xs, hist_pred_original, color='blue', linestyle='-', linewidth=2, label='Ajuste de la Red Neuronal (Datos Históricos)')

plt.title('Ajuste de la Red Neuronal a los Datos Históricos', fontsize=14)
plt.xlabel('Año del Modelo')
plt.ylabel('Vehículos Registrados')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
