```python
# Observar distribución de las columnas con nulos

columnas_a_graficar = ['Age', 'Blood Pressure', 'Cholesterol Level', 'BMI', 'Triglyceride Level']
plt.figure(figsize=(15, 6))

for i, col in enumerate(columnas_a_graficar, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df_hd[col], bins=30, kde=True, color='teal')
    plt.title(f'Distribución de {col.replace("_", " ").title()}')
    plt.xlabel(col.replace("_", " ").title())
    plt.ylabel('Frecuencia')
    plt.grid(axis='y', alpha=0.75)

plt.tight_layout()
plt.show()
```

```python
# Calcular media y desviación estándar de las columnas con nulos

media_age = df_hd['Age'].mean()
std_age = df_hd['Age'].std()

lim_sup_age = media_age + std_age
lim_inf_age = media_age - std_age

media_bp = df_hd['Blood Pressure'].mean()
std_bp = df_hd['Blood Pressure'].std()

lim_sup_bp = media_bp + std_bp
lim_inf_bp = media_bp - std_bp

media_cl = df_hd['Cholesterol Level'].mean()
std_cl = df_hd['Cholesterol Level'].std()

lim_sup_cl = media_cl + std_cl
lim_inf_cl = media_cl - std_cl

media_bmi = df_hd['BMI'].mean()
std_bmi = df_hd['BMI'].std()

lim_sup_bmi = media_bmi + std_bmi
lim_inf_bmi = media_bmi - std_bmi

media_tl = df_hd['Triglyceride Level'].mean()
std_tl = df_hd['Triglyceride Level'].std()

lim_sup_tl = media_tl + std_tl
lim_inf_tl = media_tl - std_tl


configuracion_age = [(media_age, 'blue'), (lim_sup_age, 'red'), (lim_inf_age, 'red')]
configuracion_bp = [(media_bp, 'blue'), (lim_sup_bp, 'red'), (lim_inf_bp, 'red')]
configuracion_cl = [(media_cl, 'blue'), (lim_sup_cl, 'red'), (lim_inf_cl, 'red')]
configuracion_bmi = [(media_bmi, 'blue'), (lim_sup_bmi, 'red'), (lim_inf_bmi, 'red')]
configuracion_tl = [(media_tl, 'blue'), (lim_sup_tl, 'red'), (lim_inf_tl, 'red')]
```

```python
# Observar distribución de las columnas con nulos

plt.figure(figsize=(15, 6))

plt.subplot(3, 2, 1)
sns.scatterplot(x=range(len(df_hd)), y=df_hd['Age'], s=10, color='gray')
plt.title('Desviación Estandar de Age')
plt.xlabel('Índice del registro')
plt.ylabel('Age')
for dato, color in configuracion_age:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.subplot(3, 2, 2)
sns.scatterplot(x=range(len(df_hd)), y=df_hd['Blood Pressure'], s=10, color='gray')
plt.title('Desviación Estandar de Blood Pressure')
plt.xlabel('Índice del registro')
plt.ylabel('Blood Pressure')
for dato, color in configuracion_bp:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.subplot(3, 2, 3)
sns.scatterplot(x=range(len(df_hd)), y=df_hd['Cholesterol Level'], s=10, color='gray')
plt.title('Desviación Estandar de Cholesterol Level')
plt.xlabel('Índice del registro')
plt.ylabel('Cholesterol Level')
for dato, color in configuracion_cl:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.subplot(3, 2, 4)
sns.scatterplot(x=range(len(df_hd)), y=df_hd['BMI'], s=10, color='gray')
plt.title('Desviación Estandar de BMI')
plt.xlabel('Índice del registro')
plt.ylabel('BMI')
for dato, color in configuracion_bmi:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.subplot(3, 2, 5)
sns.scatterplot(x=range(len(df_hd)), y=df_hd['Triglyceride Level'], s=10, color='gray')
plt.title('Desviación Estandar de Triglyceride Level')
plt.xlabel('Índice del registro')
plt.ylabel('Triglyceride Level')
for dato, color in configuracion_tl:
    plt.axhline(y=dato, color=color, linestyle='--')

plt.tight_layout()
plt.show()
```

```python
# Imputación numérica
# La cantidad de valores nulos no es significativa, se sustituyen por la media

df_hd['Age'] = df_hd['Age'].fillna(df_hd['Age'].median())
df_hd['Blood Pressure'] = df_hd['Blood Pressure'].fillna(df_hd['Blood Pressure'].median())
df_hd['Cholesterol Level'] = df_hd['Cholesterol Level'].fillna(df_hd['Cholesterol Level'].median())
df_hd['BMI'] = df_hd['BMI'].fillna(df_hd['BMI'].median())
df_hd['Triglyceride Level'] = df_hd['Triglyceride Level'].fillna(df_hd['Triglyceride Level'].median())
```

```python
# Calculo del IQR

df_copy = df_hd.copy() # preservar el dataset haciendo una copia

for col in columnas_numericas:

    Q1 = df_copy[col].quantile(0.25) # Primer cuartil
    Q3 = df_copy[col].quantile(0.75) # Tercer cuartil
    IQR = Q3 - Q1 # Rango Intercuartilico

    lim_inf_iqr = Q1 - 1.5 * IQR # Límite inferior
    lim_sup_iqr = Q3 + 1.5 * IQR # Límite superior

    cant_inicial = len(df_copy)
    # Filtro de los datos dentro de los límites
    df_copy = df_copy[(df_copy[col] >= lim_inf_iqr) & (df_copy[col] <= lim_sup_iqr)]
    cant_final = len(df_copy)

    print(f"Columna '{col}': Se eliminaron {cant_inicial - cant_final} registros.")

    # No hay datos atípicos
```

# Visualización de datos

```python
# Definir variables para la agrupación

cardiopatia = lambda x: 'Sin cardiopatia' if x == 'No' else 'Con Cardiopatia'
df_copy['agrupacion_card'] = df_copy['Heart Disease Status'].apply(cardiopatia)
pie_data = df_copy['agrupacion_card'].value_counts()
```

```python
# Realizar gráfico de dona

colors_pie = ['#4285F4', '#D3D3D3'] # Azul = "Con Cardiopatia", Gris = "Sin Cardiopatia"
explode_val = 0.05 # Valor de separación

# Gráfico de pastel

plt.figure(figsize=(6, 6))
wedges, texts, autotexts = plt.pie(pie_data,
                                   labels=pie_data.index,
                                   autopct='%1.1f%%',
                                   startangle=90,
                                   colors=colors_pie,
                                   textprops={'fontsize': 12, 'fontweight': 'bold'},
                                   pctdistance=0.85, # Mueve el porcentaje hacia afuera
                                   explode=(explode_val, 0)) # Resalta ligeramente la porción principal

# Círculo blanco central (estilo moderno)

theta1, theta2 = wedges[0].theta1, wedges[0].theta2
center_angle = np.deg2rad((theta1 + theta2) / 2) # Obtención el ángulo medio

cx = explode_val * np.cos(center_angle) # Se Calcula nueva posición con trigonometría
cy = explode_val * np.sin(center_angle)

centre_circle = plt.Circle((cx, cy), 0.70, fc='white') # Se dibuja el circulo
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Proporción de personas con cardiopatia', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```
