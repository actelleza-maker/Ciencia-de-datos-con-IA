```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
xs = np.arange(-1, 5, dtype=float)
ys = np.array([-4.0, 1.0, 6.0, 11.0, 16.0, 21.0], dtype=float)
print(f'xs = {xs}, ys = {ys}')
```

```python
regresion_lineal = np.polyfit(xs, ys, 1)
# y = b0x + b1
# beta --> b0 y alfa --> b1
beta, alfa = regresion_lineal
print(f'y = {round(beta, 1)}x + {round(alfa, 1)}')
```

```python
plt.plot(xs, ys, 'o', color='green', label='Datos')
plt.plot(xs, beta*xs + alfa, color='blue', label='Regresión Lineal')
plt.axhline(y = 0, color = 'red')
plt.axvline(x = 0, color = 'red')
plt.title('Say hello to the "Hello, World" of ML')
plt.xlabel('xs')
plt.ylabel('ys')
plt.legend()
plt.grid()
plt.show()
```

```python
# Perceptrón simple
capa = tf.keras.layers.Dense(units=1)
modelo = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    capa
  ])
modelo.compile(optimizer="sgd", loss="mse")
```

```python
print('Comenzando entrenamiento...')
historial = modelo.fit(xs, ys, epochs=500, verbose=True)
print('Entrenamiento finalizado')
```

```python
valor = float(input('Ingrese un valor de x para predecir y: '))
prediccion = modelo.predict(np.array([[valor]]))
print(f'El valor de y para x = {valor} es: {prediccion}')
```
